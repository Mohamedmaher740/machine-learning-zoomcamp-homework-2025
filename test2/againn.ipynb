```python
series = df.horsepower.values
print(series)
counter = int(df['horsepower'].nunique())
max_freq = 1
most_frequent_num = series[0]
for i in range(counter):
    x = series[0]
    x_freq = series[series == x].shape[0]
    print(f"{x} , {x_freq}")
    if x_freq >= max_freq:
        most_frequent_num = x
        max_freq = x_freq
    series = series[series != x]

print(f"max frequency = {max_freq} and most frequent number is: {most_frequent_num}")
# x = series[0]
# series = series[series != x]
# series[0]
```
