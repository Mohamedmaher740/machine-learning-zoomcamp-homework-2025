## Q6

```python
initial_median = df.horsepower.median()
uniques = df.horsepower[~df.horsepower.isnull()].unique()
max_freq = 1
for num in uniques:
    num_freq = list(df.horsepower).count(num)
    if num_freq > max_freq:
        max_freq = num_freq
        most_frequrnt_num = num
print(f"Most frequent num is: {most_frequrnt_num} | Max frequency = {max_freq}")
df2 = df.copy()
df2.horsepower = df2.horsepower.fillna(most_frequrnt_num)
final_median = df2.horsepower.median()
print(f"Median before fill null values = {initial_median} and after = {final_median}")
if final_median > initial_median:
    print("So, Median increased")
elif initial_median > final_median:
    print("So, Median decreased")
else:
    print("So, Median is constant")
```
