```python
initial_median = df.horsepower.median()
initial_median
```

```python
row_series = df.horsepower[df.horsepower.isnull() == False] # series without null
unique_values = row_series.nunique()
ser = row_series.values # convert series to array
ser
```

```python
max_freq = 1
MFN = ser[0]
for i in range(unique_values):
    x = ser[0]
    x_freq = ser[ser == x].shape[0]
    # print(f"{x} , {x_freq}")
    if x_freq >= max_freq:
        MFN = x
        max_freq = x_freq
    ser = ser[ser != x]
    
MFN
```

```python
df1 = df.copy()
df1.horsepower = df1.horsepower.fillna(MFN)
df1.horsepower.median()
```
