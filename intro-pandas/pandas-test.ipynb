# Pandas Experiments

```python
import pandas as pd
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/alexeygrigorev/datasets/master/car_fuel_efficiency.csv")
```

```python
initial_median = df.horsepower.median()
series = df.horsepower[df.horsepower.isnull() == False].values
counter = int(df['horsepower'].nunique())
max_freq = 1
most_frequent_num = series[0]
for i in range(counter):
    x = series[0]
    x_freq = series[series == x].shape[0]
    if x_freq >= max_freq:
        most_frequent_num = x
        max_freq = x_freq
    series = series[series != x]

df.horsepower.fillna(most_frequent_num)
final_median = df.horsepower.median()

print(f"inital_median = {initial_median} , final_median = {final_median}")
```
